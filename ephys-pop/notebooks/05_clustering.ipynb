## 1. Clustering: fitting blobs to points

**Clustering** means: given a pile of points with no labels, discover the groups.
Here is such a pile — two features measured for a few hundred points, and nothing
else:

```python
rng = np.random.default_rng(0)
centres = np.array([[0, 0], [5, 4], [8, -1]])
toy = np.vstack([rng.normal(c, 0.8, size=(120, 2)) for c in centres])

plt.figure(figsize=(5, 4))
plt.scatter(toy[:, 0], toy[:, 1], s=12, color="0.5")
plt.title("a pile of unlabelled points"); plt.xlabel("feature 1"); plt.ylabel("feature 2")
plt.show()
```

You can *see* three groups. The job is to get a computer to see them too, with no
labels to learn from. Our tool is the **Gaussian mixture**.

- A **Gaussian** is a blob of points scattered around a **centre**, densest in the
  middle and thinning out with distance — a 2-D bell curve. Its **spread** can be
  wider in one direction than another, so the blob is an ellipse.
- A **mixture** is several such blobs at once, each with its own centre, spread, and
  share of the points.
- **Fitting** a mixture to the data means adjusting those centres and spreads until
  the blobs sit over the data as well as possible. Once fitted, every point can be
  **assigned** to the blob it most likely came from — that's the clustering.

```python
from matplotlib.patches import Ellipse

def draw_blob(ax, mean, cov, color, label=None):
    # Shade a Gaussian blob as its 1-sigma and 2-sigma ellipses, with an x at the centre.
    vals, vecs = np.linalg.eigh(cov)
    angle = np.degrees(np.arctan2(vecs[1, 1], vecs[0, 1]))
    for k, alpha in [(2, 0.15), (1, 0.35)]:
        ax.add_patch(Ellipse(mean, 2 * k * np.sqrt(vals[1]), 2 * k * np.sqrt(vals[0]),
                             angle=angle, color=color, alpha=alpha, lw=0))
    ax.plot(*mean, "x", color="k", ms=7, mew=1.5, label=label)

blobs = [(np.array([0.0, 0.0]), np.array([[1.0, 0.6], [0.6, 1.0]])),
         (np.array([5.0, 4.0]), np.array([[0.5, 0.0], [0.0, 1.2]])),
         (np.array([8.0, -1.0]), np.array([[1.2, -0.4], [-0.4, 0.5]]))]

fig, ax = plt.subplots(1, 3, figsize=(13, 3.8), sharex=True, sharey=True)
# (1) one Gaussian: a centre and a spread
draw_blob(ax[0], *blobs[0], "tab:blue")
ax[0].annotate("centre", blobs[0][0], xytext=(2.5, 2.5), arrowprops=dict(arrowstyle="->"))
ax[0].annotate("spread (1σ, 2σ)", (1.6, -1.2), xytext=(3.5, -2.8), arrowprops=dict(arrowstyle="->"))
ax[0].set_title("a Gaussian: centre + spread")
# (2) a mixture: several blobs, each with its own centre and spread
for (m, c), g in zip(blobs, range(3)):
    draw_blob(ax[1], m, c, ps.plotting.unit_color(g))
ax[1].set_title("a mixture: several blobs")
# (3) assignment: a point belongs to the blob it most likely came from
for (m, c), g in zip(blobs, range(3)):
    draw_blob(ax[2], m, c, ps.plotting.unit_color(g))
q = np.array([4.2, 2.2])
ax[2].plot(*q, "o", color="k", ms=8); ax[2].text(q[0] + 0.3, q[1] - 0.2, "which blob?", fontsize=9)
for m, _ in blobs:
    ax[2].plot([q[0], m[0]], [q[1], m[1]], ":", color="0.4", lw=1)
ax[2].set_title("assign each point to its most likely blob")
for a in ax:
    a.set_xlim(-3.5, 11.5); a.set_ylim(-4, 7); a.set_xlabel("feature 1")
ax[0].set_ylabel("feature 2")
plt.tight_layout(); plt.show()
```

Now fit a three-blob mixture to the pile. `sklearn` does the fitting; we ask it for
each point's most likely blob, and draw the blobs it found over the points:

```python
from sklearn.mixture import GaussianMixture

gmm = GaussianMixture(n_components=3, random_state=0).fit(toy)
assignment = gmm.predict(toy)                  # each point's most likely blob

plt.figure(figsize=(5, 4))
ax = plt.gca()
for g in range(3):
    m = assignment == g
    ax.scatter(toy[m, 0], toy[m, 1], s=12, color=ps.plotting.unit_color(g))
    draw_blob(ax, gmm.means_[g], gmm.covariances_[g], ps.plotting.unit_color(g))
plt.title("the fitted mixture, and each point's blob")
plt.xlabel("feature 1"); plt.ylabel("feature 2"); plt.show()
```

The mixture coloured the three groups correctly, with no labels to learn from. Our
spikes are just such a pile of points — in the **depth–amplitude feature space** from
Notebook 4 — and each neuron is a blob. Two questions remain: *how many* blobs are
there, and *where does each spike belong?*

## 2. How many units? The BIC

Now the real thing. Here are our spikes in the **depth–amplitude feature space** from
Notebook 4 — the pile we have to cluster, with no labels:

```python
plt.figure(figsize=(5.5, 4.5))
plt.scatter(depth, amplitude, s=10, color="0.5")
plt.title(f"{len(depth)} detected spikes, unlabelled")
plt.xlabel("depth (µm)"); plt.ylabel("amplitude (µV)"); plt.show()
```

On the toy we told the mixture to look for three blobs. Here, how many? You can count
clouds by eye, but a computer can't, and on messier data neither can you — the
number of units is part of what we're trying to discover. First, see what goes wrong
when you guess:

```python
fig, ax = plt.subplots(1, 3, figsize=(13, 4), sharex=True, sharey=True)
for a, k in zip(ax, [3, 6, 9]):
    lab_k = GaussianMixture(k, n_init=5, random_state=0).fit_predict(features)
    for g in range(k):
        a.scatter(depth[lab_k == g], amplitude[lab_k == g], s=10, color=ps.plotting.unit_color(g))
    a.set_title(f"k = {k}  ({'too few' if k == 3 else 'just right' if k == 6 else 'too many'})")
    a.set_xlabel("depth (µm)")
ax[0].set_ylabel("amplitude (µV)")
plt.tight_layout(); plt.show()
```

With **k = 3** (left) the mixture is forced to lump distinct units into one blob — a
poor fit, with points stranded far from their cluster's centre. With **k = 9** (right)
it splits a *single* real unit across two colours — which explains the data no better
than one cluster did, since those points were already well accounted for. **k = 6**
(middle) is the sweet spot: one cluster per real group, none wasted.

The **BIC (Bayesian Information Criterion)** turns this trade-off into one number to
minimise: $\mathrm{BIC} = \text{misfit} + \text{penalty}$, where *misfit* measures how
badly the model explains the data (smaller is better) and *penalty* charges a fixed
cost for every cluster's parameters. **Why does it have an elbow?** Each cluster you add
*up to* the true number lets the model wrap a Gaussian around a real group it was
missing, so the misfit drops **steeply** — easily worth the small penalty. Each cluster
*beyond* the true number only carves up a group that was *already* well explained, so
the misfit barely moves while you keep paying the penalty. A steep drop, then a flat
plateau — and the bend between them, the **elbow, is the honest unit count.**

<details>
<summary><b>▸ Go deeper: Gaussian mixtures, and where the BIC comes from (optional)</b></summary>

**The model.** A Gaussian mixture says each spike's feature vector $x$ was drawn from
one of $k$ Gaussians — one per unit — with unknown mixing weights $\pi_j$, means
$\mu_j$, and covariances $\Sigma_j$:

$$p(x \mid \theta) = \sum_{j=1}^{k} \pi_j\, \mathcal{N}(x;\, \mu_j, \Sigma_j).$$

We fit $\theta$ by maximising the log-likelihood $\log \hat L = \sum_i \log p(x_i\mid\theta)$,
using the **EM algorithm**: alternate between the **E-step** (compute each spike's
*responsibility* — the posterior probability it belongs to unit $j$) and the
**M-step** (re-estimate each $\pi_j, \mu_j, \Sigma_j$ as responsibility-weighted
averages). This is a soft version of the assign-then-average loop you already know.
(k-means is the special case with equal spherical $\Sigma_j = \sigma^2 I$ and *hard*
responsibilities — which is exactly why it merges close clusters and splits elongated
ones, and why the full mixture does better here.)

**The penalty.** More clusters always fit better, so raw likelihood can't choose $k$.
The BIC adds a complexity penalty:

$$\mathrm{BIC}(k) = -2\log \hat L \;+\; m_k \log n,$$

with $m_k$ the number of free parameters and $n$ the number of spikes. It is a
large-$n$ approximation (via **Laplace's method**) to $-2\log p(\text{data}\mid k)$,
the **model evidence** — the probability the data would arise under a $k$-unit model,
with the parameters averaged out rather than fitted. Extra clusters beyond the truth
buy almost no likelihood but keep paying $\log n$ per parameter, so the curve
flattens: the **elbow is the honest unit count.**
</details>

The six clusters match the six blobs by eye. But we clustered on just two features
(depth, amplitude) — how do we know these are *really* six distinct neurons, and not
(say) one neuron artificially split? A good check is to look at the clusters in the
**full waveform space**: every spike's complete multichannel snippet (~2000 numbers),
where two genuinely different neurons should differ in *many* ways at once.

We can't plot 2000 dimensions, so we squash them to 2-D — and *which* squashing matters.
The obvious tool is **PCA**, but PCA is **linear**: it keeps the two directions of
largest variance, and distinct clusters can still land on top of each other in that
projection (you saw exactly this in Notebook 4, where PC1 just re-encoded amplitude).

**t-SNE** is a **nonlinear** embedding built for this job. It gives up on preserving
distances and preserves only **neighbourhoods**:

- **Step 1, in the original space:** for every point, find its nearest neighbours (a
  few dozen; the number is the `perplexity` knob) and note how close each one is.
  That's all t-SNE ever looks at: each point's list of neighbours. A point on the far
  side of the space doesn't exist as far as this point is concerned.
- **Step 2, on a 2-D sheet:** drop the points anywhere, then nudge them repeatedly:
  **pull** each point towards its neighbours from the list, **push** it away from
  everything else. Stop when every point sits next to the same neighbours it had in
  the original space.

Because only neighbour lists matter, a group of points that was separate in the
original space ends up as a separate **island**, however the dimensions were tangled.
But the same fact is the catch: *how far apart* islands land, and *how big* they are,
comes from the pushing and nudging, not from the data. Watch it on toys where we know
the answer:

```python
rng = np.random.default_rng(1)
# toy A: two tight blobs close together, one loose blob far away
A = np.vstack([rng.normal([0, 0], 0.3, (100, 2)), rng.normal([1.5, 0], 0.3, (100, 2)),
               rng.normal([12, 0], 1.5, (100, 2))])
lab_A = np.repeat([0, 1, 2], 100)
# toy B: one featureless Gaussian cloud — no clusters at all
B = rng.normal(0, 1, (300, 2))

fig, ax = plt.subplots(1, 4, figsize=(15, 3.6))
# (1) toy A, with one point's neighbour list drawn
for g in range(3):
    ax[0].scatter(*A[lab_A == g].T, s=10, color=ps.plotting.unit_color(g))
i = 230; d = np.linalg.norm(A - A[i], axis=1); nn = np.argsort(d)[1:16]   # a point and its 15 nearest
for j in nn:
    ax[0].plot([A[i, 0], A[j, 0]], [A[i, 1], A[j, 1]], color="k", lw=0.6, alpha=0.7)
ax[0].plot(*A[i], "ko", ms=6)
ax[0].set_title("toy A: two close blobs, one far loose blob\n(one point's neighbour list drawn)", fontsize=9)
# (2) its t-SNE
emb_A = ps.tsne_embedding(A, perplexity=30)
for g in range(3):
    ax[1].scatter(*emb_A[lab_A == g].T, s=10, color=ps.plotting.unit_color(g))
ax[1].set_title("t-SNE of toy A: three islands —\nbut equally spaced, equally sized", fontsize=9)
# (3) toy B and (4) its t-SNE
ax[2].scatter(*B.T, s=10, color="0.5")
ax[2].set_title("toy B: one cloud, no structure", fontsize=9)
emb_B = ps.tsne_embedding(B, perplexity=10)
ax[3].scatter(*emb_B.T, s=10, color="0.5")
ax[3].set_title("t-SNE of toy B: lumps appear\nthat aren't in the data", fontsize=9)
for a in ax[1:]:
    a.set_xticks([]); a.set_yticks([])
plt.tight_layout(); plt.show()
```

Toy A: the two close blobs and the far blob come out as **three islands**, which is
what we want — neighbourhoods were preserved. But the far blob is no longer far, and
the loose blob is no longer loose: distances and sizes in the map mean nothing. Toy B:
a single structureless cloud comes out lumpy, because pulling neighbours together and
pushing everyone else apart makes lumps whether or not any were there.

So why not cluster with t-SNE all the time? Because it is a **picture, not a model**.
There are no blobs with centres and spreads to read off, no likelihood, so no BIC to
choose the number of units; a new spike can't be placed on the map without rerunning
everything; the layout changes from run to run; it is slow for millions of spikes; and
it can draw clusters that aren't there. We cluster in a space whose distances we
understand, and use t-SNE afterwards to *look*: if the clusters we found land in
separate islands, that's reassuring. Compare PCA and t-SNE on our snippets:

```python
from sklearn.decomposition import PCA
X = snippets.reshape(len(snippets), -1)          # every spike as its full multichannel snippet
pca2 = PCA(n_components=2).fit_transform(X)       # linear 2-D projection
emb = ps.tsne_embedding(X)                         # nonlinear 2-D embedding (sklearn's TSNE)

fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))
for u in np.unique(labels):
    m = labels == u
    axes[0].scatter(pca2[m, 0], pca2[m, 1], s=10, color=ps.plotting.unit_color(int(u)))
    axes[1].scatter(emb[m, 0], emb[m, 1], s=10, color=ps.plotting.unit_color(int(u)))
axes[0].set_title("PCA (linear) — some clusters overlap")
axes[1].set_title("t-SNE (nonlinear) — clean islands")
for a in axes:
    a.set_xticks([]); a.set_yticks([])
plt.tight_layout(); plt.show()
```

In the **PCA** projection several clusters pile up in the middle — the linear map can't
pull them apart. In the **t-SNE** map all six sit in clean, separate islands: strong
confirmation that our six clusters really are six distinct units. That's why t-SNE is a
staple of spike-sorting QC.

Remember the toys, though: the gaps between islands and their sizes mean nothing, and
lumps can appear on their own. Use t-SNE to eyeball separation, never as proof on its
own.

## 5. Did we recover the real units?

Because this is synthetic data, we can check. Match each learned template to the
true template it most resembles and compare their shapes — they should be nearly
identical, up to noise.

```python
gt = rec.ground_truth
corr = np.zeros((len(templates), gt.n_units))
for a in range(len(templates)):
    for b in range(gt.n_units):
        corr[a, b] = np.corrcoef(templates[a].ravel(), gt.templates[b].ravel())[0, 1]
best = corr.max(axis=1)
print("each learned template's correlation to its best-matching true template:")
print(np.round(best, 3))
print(f"worst match: {best.min():.3f}  ->  all six units recovered" if best.min() > 0.8
      else f"worst match: {best.min():.3f}")
```
